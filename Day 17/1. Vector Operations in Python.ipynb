# Importing Dependencies

```python
import numpy as np
import pandas as pd
```

# Dot Product of 2 Vectors

```python
a = np.array([2,3])

b = np.array([4,4])

dot_product = np.dot(a,b)
```

```python
print(dot_product)
```

```python
c = np.array([40,20,35])
d = np.array([53,24,68])

dot_product = np.dot(c,d)
```

```python
print(dot_product)
```

# Cross Product of 2 Vectors

```python
a = np.array([2,3])

b = np.array([4,4])

cross_product = np.cross(a,b)
```

```python
print(cross_product)
```

```python
c =  np.array([5,10,20])  # 3D x,y,z or i,j,k

d = np.array([18,32,50])

cross_product = np.cross(c,d)
```

```python
print(cross_product)
```
