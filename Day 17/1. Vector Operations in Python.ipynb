# Dot Product of 2 Vectors

# Cross Product of 2 Vectors

```python
a = np.array([2,3])

b = np.array([4,4])

cross_product = np.cross(a,b)
```

```python
print(cross_product)
```

```python
c =  np.array([5,10,20])  # 3D x,y,z or i,j,k

d = np.array([18,32,50])

cross_product = np.cross(c,d)
```

```python
print(cross_product)
```

# Projection of "a" vector on "v" vector 

```python
a = np.array([2, 5])

v = np.array([8, -6])

# Magnitude of V vector:

magnitude_of_v = np.sqrt(sum(v**2))   # Magnitude is √((v1^2)+(v2^2))
dot_product_of_a_and_v = np.dot(a,v)
proj_of_a_on_v = (dot_product_of_a_and_v/magnitude_of_v**2)*v
print("Projection of a vector on v: ",proj_of_a_on_v)
```

```python
# 3D Vector

a = np.array([23, 45, 62])  # Shape of both vector should be same

v = np.array([45, 82, 67])

# Magnitude of V vector:

magnitude_of_v = np.sqrt(sum(v**2))   # Magnitude is √((v1^2)+(v2^2))
dot_product_of_a_and_v = np.dot(a,v)
proj_of_a_on_v = (dot_product_of_a_and_v/magnitude_of_v**2)*v
print("Projection of a vector on v: ",proj_of_a_on_v)
```
