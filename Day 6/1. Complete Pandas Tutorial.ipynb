```python
# Importing the pandas library

import pandas as pd
import numpy as np
```

# Creating a Pandas DataFrame

```python

# Importing the boston House price data
from sklearn.datasets import fetch_openml

# Load the Boston Housing dataset from OpenML
boston_dataset = fetch_openml(name='boston', version=1, as_frame=True)

print(boston_dataset)
```

```python
type(boston_dataset)
```

```python
# Pandas DataFrame

boston_df = pd.DataFrame(boston_dataset.data, columns=boston_dataset.feature_names)
```

```python
type(boston_df)
```

```python
print(boston_df)
```

```python
boston_df.head()
```

# Statistical Measures

```python
# Count or number of values

boston_df.count()
```

```python
# Mean value - col wise (excluding categorical columns)
boston_df.select_dtypes(include=[np.number]).mean()
```

```python
# Standard Deviation col wise

boston_df.select_dtypes(include=[np.number]).std()
```

```python
# Minimum values in each col

boston_df.select_dtypes(include=[np.number]).min()
```

```python
# Maximum values in each col

boston_df.select_dtypes(include=[np.number]).max()
```

```python
# All satistical measures in the DataFrame

boston_df.describe()
```
