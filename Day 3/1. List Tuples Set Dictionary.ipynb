Types of Objects in Python:
1. <u>Immutable Objects</u> - Immutable objects are those whose values cannot be changed once its created
2. <u>Mutable Objects</u> - Mutable objects are those whose values can be changed once its created

Immutable Objects:
1. int
2. float
3. string
4. bool
5. tuple

Mutable Objects:
1. List
2. Set
3. Dictionary
