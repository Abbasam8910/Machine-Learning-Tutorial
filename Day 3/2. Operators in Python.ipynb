# Arithmatic Operators

```python
num1=20
num2=30

# Addition
add = num1+ num2

# Subtraction
sub = num1 - num2

# Multiplication
mul = num1 * num2

# Division
div = num1/num2

# Exponent
exp = num1 ** num2

# Modulus
mod = num1 % num2

print("Sum = ", add)
print("Difference = ", sub)
print("Multiplication =", mul)
print("Division = ", div)
print("Exponent = ", exp)
print("Modulus = ", mod)
```

# Assignment Operators

- +=
- -=
- *=

- %=

```python
a = 5
print(a)
```

```python
a = 5

a +=5 # a = a+ 5

print(a)
```

```python
b=12

b -=2 # b = b-2

print(b)
```
