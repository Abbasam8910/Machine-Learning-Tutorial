Convertion of one data type to another

```python
# int to float

x = 10
print(x)
type(x)
```

```python
y=float(x)

print(y)
type(y)
```

```python
# Float to int

x = 2.3
print(x)
type(x)
```

```python
y = int(x)
print(y)
type(y)
```

Slicing

```python
my_string = "Programming"

print(my_string[1:5]) # Print all the letters between index 0 and index 4 

# Values from index 1 to 5-1 = 4 will be sliced
```

```python
# step 

print(my_string[0:11:2]) # Step of 2 ie every 2nd letter will be removed
```
