# Importing Required Dependencies

# Seperating The Features and Target

# GridSearchCV

GridSearchCV is used for determining the Best parameters for our model

```python
# Load our svc model

model = SVC()
```

```python
parameters = {'kernel':['linear','poly','rbf','sigmoid'], 
              'C':[1,5,10,20]}
```

```python
# Grid Search

classifer = GridSearchCV(model, parameters, cv=5)
```

```python
# Fitting the data in our model
classifer.fit(X, Y)
```

```python
classifer.cv_results_
```

```python
# Best parameters 

best_parameters = classifer.best_params_
print(best_parameters)
```

```python
# High Accuracy

high_accuracy = classifer.best_score_
print(high_accuracy)
```

```python
# Loading the Result to pandas DataFrame

result = pd.DataFrame(classifer.cv_results_)
```
