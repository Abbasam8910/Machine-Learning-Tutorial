# Matrix Addition

Two matrices can be only added if they have same shape

```python
# Creating two matrices

matrix_1 = np.array([[3,5], [6,7]])
matrix_2 = np.array([[5,7], [1,2]])

```

```python
print(matrix_1)
```

```python
print(matrix_2)
```

```python
matrix_1.shape
```

```python
matrix_2.shape
```

```python
# Adding Two Matrices

sum = matrix_1 + matrix_2
```

# Matrix Subtraction

```python
print(matrix_1)
```
