# Importing Required Libraries

```python
import numpy as np
import pandas as pd
```

# Matrix Addition

Two matrices can be only added if they have same shape

```python
matrix_1 = np.array([[3,5], [6,7]])
matrix_2 = np.array([[5,7], [1,2]])

matrix_add = matrix_1 + matrix_2

print(matrix_add)
```

```python
matrix_add.shape
```

```python
print(np.add(matrix_1, matrix_2))
```

```python
#   Create two matrices with random values

mat1= np.random.randint(100, size=(3,3))
mat2 = np.random.randint(100, size=(3,3))
print(mat1+mat2)
```

# Matrix Subtraction

```python
matrix_1 = np.array([[3,5], [6,7]])
matrix_2 = np.array([[5,7], [1,2]])

matrix_sub = matrix_1 - matrix_2

print(matrix_sub)
```

```python
matrix_sub.shape
```

```python
print(np.subtract(matrix_1, matrix_2))
```

# Matrix Multiplication

```python
matrix_1 = np.array([[3,5], [6,7]])
matrix_2 = np.array([[5,7], [1,2]])

matrix_mult = matrix_1 * matrix_2

print(matrix_mult)
```

```python
print(np.multiply(matrix_1, matrix_2))
```
