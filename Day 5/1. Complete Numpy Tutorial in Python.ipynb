# List Vs Numpy - Time Taken

```python
from time import process_time
```

TIme Taken by a List

```python
python_list = [i for i in range(100000000)] # Inset values to list from 0 to 10000

start_time = process_time()

python_list = [i+5 for i in python_list]

end_time = process_time()

print(end_time - start_time)
```

```python
np_array = np.array([i for i in range(100000000)])

start_time = process_time()

np_array +=5

end_time = process_time()

print(end_time - start_time)
```

Numpy is Faster

# Initial Placeholders in Numpy Array

Inital values prenset in the Numpy Array

```python
# create a numpy array of Zero

x = np.zeros((4,5))
print(x)
type(x)
```

```python
# Create a numpy array of ones

y = np.ones((3,3))
print(y)
```

```python
# Array of a particular value

z = np.full((5,4), 5)

print(z)
```

```python
# Create an identity matrix

a = np.eye((5))
print(a)
```

```python
# Create a numpy array with random values

b = np.random.random((3,4))
b = np. random.randint((3,4))
print(b)
```

```python
c = np. random.randint(10,100,(3,4))
print(c)
```

```python
# Create array of evenly spaced values -> specifying the number of values required

d = np.linspace(10,30,5)
print(d)
```

```python
# Array of evenly spaced values -> specifying the step

e = np.arange(10,30,3)
print(e)
```

```python
# Convert a list to an numpy array

list2 = [10,11,12,14,15]
print(type(list2))
np_array = np.asarray(list2)
print(np_array)
print(type(np_array))
```
