### Factorial of a Number

Factorial of a number is the product of all the positive intergers less than or equal to the given number

Factorial of 5 = 5 * 4 * 3 * 2 * 1 = 120

```python
number = int(input("Enter a number to check its factorial"))

factorial = 1

if number == 0:
    print("Factorial of 0 is 1")
else:
    for i in range(1, number+1): # range (1,6) --> 1,2,3,4  [ie it will include fist value and exclude the last value]
        factorial = factorial * i
    print(f"Factorial of the {number} is: ", factorial)
```

### Factorial Function

```python
# Lets create using a function

def factorial_value(num):
    factorial=1
    
    if num == 0:
        return factorial
    else:
        for i in range(1, num+1):
            factorial = factorial*i
        return factorial
        
```

```python
print(factorial_value(5))
```

```python
print(factorial_value(10))
```
