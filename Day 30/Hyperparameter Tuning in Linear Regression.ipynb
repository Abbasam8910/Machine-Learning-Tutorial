# Importing Requied Dependencies

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

# Preprocessing

```python
car_data = pd.read_csv('cars.csv')
```

```python
# first five rows

car_data.head()
```

```python
# Checking null values

car_data.isnull().sum()
```

# Linear Regression

```python
# Splitting Feature and Target

X = car_data[["Year", "Kilometers_Driven", "Mileage", 
             "Engine", "Power", "Seats"]]

Y = car_data['Price']
```

```python
print(X)
```

```python
print(Y)
```

```python
# Split the Training and Test data

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, random_state=0, test_size=0.2)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, Y_train)
```

```python
prediction = model.predict(X_test)
```

```python
r2_score = r2_score(prediction, Y_test)
```

```python
print(r2_score*100)
```

Our model's accuracy is very less therefore now we will try hyperparameter tuning and check wether we are getting better results or not.

```python
model.get_params()
```

# Linear Regression Model Using GridSearchCV

```python
model = LinearRegression()
```

```python
parameters = {'copy_X': [True, False],
 'fit_intercept': [True, False],
 'n_jobs': [1,5,10,15,None],
 'positive':  [True,False],
}
```

```python
# Grid Search

predictor = GridSearchCV(model, parameters, cv=5)
```

```python
predictor.fit(X, Y)
```

```python
predictor.cv_results_
```

```python
# Best Parameters

best_parameters = predictor.best_params_
print(best_parameters)
```

```python
# High Accuracy 

high_accuracy = predictor.best_score_
print(high_accuracy)
```

#  Linear Regression Model Using RandomizedSearchCV

```python
model = LinearRegression()
```

```python
model.get_params()
```

```python
parameters = {'copy_X': [True,False], 
               'fit_intercept': [True,False], 
               'n_jobs': [1,5,10,15,None], 
               'positive': [True,False]}
```

```python
# Grid search

predictor  = RandomizedSearchCV(model, parameters,n_iter=1000, cv=5)
```

```python
predictor.fit(X, Y)
```

```python
# Best parameters

best_parameters = predictor.best_params_
print(best_parameters)
```

```python
# High accuracy 

high_accuracy = predictor.best_score_
print(high_accuracy)
```
