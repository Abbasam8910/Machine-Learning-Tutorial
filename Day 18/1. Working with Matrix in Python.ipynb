# Creating Matrices with Random Values

```python
random_matrix = np.random.rand(3,3)
print(random_matrix)
```

```python
# Creating Matrix with random integers
random_matrix_int = np.random.randint(100, size=(3,3)  )
print(random_matrix_int)
```

# Matrix With all the values 1

```python
matrix_3 = np.ones((3,3))
print(matrix_3)
```

```python
# All values integers

matrix_3 = np.ones((3,3), dtype=int)
print(matrix_3)
```

# Null Matrix or Zero Matrix

```python
null_matrix = np.zeros((3,3))
print(null_matrix)
```

```python
# Null Matrix integers

null_int_matrix = np.zeros((4,4), dtype=int)
print(null_int_matrix)
```

# Identity Matrix

```python
identity_matrix = np.eye(4,4)
print(identity_matrix)
```

```python

identity_matrix = np.eye(6, 6)
print(identity_matrix)
```

# Transpose of a Matrix

```python
mat_A = np.random.randint(100, size=(3,3))
print(mat_A)
```

```python
transpose = np.transpose(mat_A)
print(transpose)
```
