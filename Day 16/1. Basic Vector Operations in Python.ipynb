# Importing Dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set()  # To load basic default themes for our charts 
```

# Plotting a Vector

```python
plt.quiver(0,0,4,5)
plt.show()
```

Our plot is not plotting correctly so we will mention x and y properly

```python
plt.quiver(0,0, 4,5, scale_units = 'xy', angles = 'xy', scale =1, color='r') # (0,0) -> base point
plt.quiver(0,0, -3,-6, scale_units = 'xy', angles = 'xy', scale =1, color='y') 
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

# Addition of Two Vectors

```python
vector_1 = np.asarray([0,0, 2,3])
vector_2 = np.asarray([0,0, 3, -2])

sum = vector_1 + vector_2

print(sum)
```

```python
#  Ploting above three vectors

plt.quiver(0,0, 2,3, scale_units = 'xy', angles = 'xy', scale =1, color='r') # (0,0) -> base point
plt.quiver(0,0, 3,-2, scale_units = 'xy', angles = 'xy', scale =1, color='y') 
plt.quiver(0,0, 5,1, scale_units = 'xy', angles = 'xy', scale =1, color='b')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```
