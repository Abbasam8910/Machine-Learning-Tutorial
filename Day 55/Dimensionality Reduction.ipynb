Dimensionality Reduction:

1. Feature Selection
2. Feature Extraction - Principal Component Analysis (PCA)

```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
X, Y = make_blobs(n_samples=500, n_features=5, centers=3, cluster_std=1.5, random_state=42)
```

```python
print(X)
```

```python
print(Y)
```

```python
scalar = StandardScaler()
```

```python
X_scaled = scalar.fit_transform(X)
```

```python
print(X_scaled)
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
X_pca
```

```python
df_pca = pd.DataFrame(X_pca, columns=['PCA1', 'PCA2'])
df_pca['labels'] = Y
```

```python
df_pca.head()
```

```python
sns.scatterplot(data=df_pca, x='PCA1', y='PCA2', hue='labels', palette='Set2')
```
