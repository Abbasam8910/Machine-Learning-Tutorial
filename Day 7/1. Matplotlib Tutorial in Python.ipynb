Matplotlib:

- Useful for making plots

```python
# Importing matplotlib library
import matplotlib.pyplot as plt
```

```python
# Import Numpy library to get data for our plots
import numpy as np
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)
z = np.cos(x)
```

```python
print(x)
```

```python
print(y)
```

```python
print(z)
```

# Plotting the Data

```python
# sine wave
plt.plot(x,y)
plt.show()
```

```python
# Cosine wave
plt.plot(x,z)
plt.show()
```

```python
# Adding title x-axis and y-axis labels

plt.plot(x,y)
plt.xlabel('Angle')
plt.ylabel('Sine Value')
plt.title('Sine Wave')
plt.show()
```

```python
# Parabola

x = np.linspace(-10,10,20)
y = x **2

plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, 'g+')
plt.show()
```

```python
plt.plot(x,y, 'rx')
```

```python
x = np.linspace(-5,5,50)
plt.plot(x, np.sin(x), 'g-')
plt.plot(x, np.cos(x), 'r--')
plt.show()
```

# Bar Plot

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
languages = ['English', 'French', 'Spanish', 'Germen', 'Latin']
people = [100, 50, 150, 40, 70]
ax.bar(languages,people)
plt.xlabel('languages')
plt.ylabel('people')
plt.title('languages vs People')
plt.show()
```

# Pie Chart

```python
fig1 = plt.figure()
ax = fig1.add_axes([0,0,1,1])
languages = ['English', 'French', 'Spanish', 'Germen', 'Latin']
people = [100, 50, 150, 40, 70]
ax.pie(people, labels=languages, autopct='%1.1f%%')
plt.show()
```

# Scatter Plot

```python
x = np.linspace(0,10,30)
y = np.sin(x)
z = np.cos(x)

fig2 = plt.figure()
ax = fig2.add_axes([0,0,1,1])
ax.scatter(x,y, color='g')
ax.scatter(x,z, color='b')
plt.show()
```
