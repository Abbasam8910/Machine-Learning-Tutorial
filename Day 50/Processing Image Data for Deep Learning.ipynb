**Topics:**

- Reading an image file and converting it to a numpy array
- Resizing an image
- RGB to Grayscale conversion

```python
# Get an image using web get

import requests

url = "https://tractive.com/blog/wp-content/uploads/2016/04/puppy-care-guide-for-new-parents.jpg"
response = requests.get(url)

# Save the image
with open("puppy.jpg", "wb") as f:
    f.write(response.content)

print("Image downloaded!")
```

**Libraries that can be used for image processing:**

- matplotlib.image
- Pillow
- OpenCV (cv2)

```python
# Importing image modules from matplotlib library

import matplotlib.image as mpimg
import matplotlib.pyplot as plt
```

```python
# Loading an image through matplotlib.image module

img = mpimg.imread('puppy.jpg')
```

```python
type(img)
```

```python
print(img.shape)
```

```python
print(img)
```

```python
# Display the image from numpy array

img_plot = plt.imshow(img)
plt.show()
```

# Resizing the Image using Pillow Library

```python
from PIL import Image
```

```python
img = Image.open('puppy.jpg')
img_resized = img.resize((200,200))
```

```python
img_resized.save('dog_image_resized.jpg')
```

```python
# Display the image from numpy array

img_res = mpimg.imread('dog_image_resized.jpg')
img_res_plot = plt.imshow(img_res)
plt.show()
```

```python
print(img_res.shape)
```

# Converting RGB images to Grayscale image using OpenCV

```python
# Importing OpenCV library

import cv2
```

```python
img = cv2.imread('puppy.jpg')
```

```python
type(img)
```

```python
img.shape
```

```python
grayscale_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
type(grayscale_image)
```

```python
grayscale_image.shape
```


cv2.imshow() will display the image. 

```python
cv2.imshow('Grayscale Image', grayscale_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plt.imshow(grayscale_image, cmap='gray')
plt.show()
```

```python
# Saving Grayscale image

cv2.imwrite('dog_grayscale_image.jpg', grayscale_image)
```
