# Importing Library

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

# Seaborn Build in Dataset

Seaborn has some Build in Datasets

```python
# Total bill vs tip dataset

tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
tips.dtypes
```

```python
# Setting a theme for plots

sns.set_theme()
```

```python
# Visualize the data

sns.relplot(data=tips, x = 'total_bill', y = 'tip', col='time', hue='smoker', style='smoker', size='size' )
```

```python
# House price dataset

from sklearn.datasets import fetch_openml

# Load the Boston Housing dataset from OpenML
boston_dataset = fetch_openml(name='boston', version=1, as_frame=True)
```

```python
# Loading our boston_dataset into pandas DataFrame

boston_df = pd.DataFrame(boston_dataset.data, columns=boston_dataset.feature_names)
boston_df['PRICE'] = boston_dataset.target
```

```python
print(boston_dataset)
```

```python
boston_df.head()
```

# Distribution Plot


```python
sns.displot(boston_df["PRICE"])
```

```python
sns.distplot(boston_df["PRICE"])
```

# Correlation:

- Postive Correlation
- Negative Correlation

# Heat Map

```python
correlation = boston_df.corr()
correlation.head()
```

```python
# Constructing Heat Map
plt.figure(figsize=(10,10))
sns.heatmap(correlation, cbar=True, square=True, fmt='.1f',  annot=True, annot_kws={'size':8}, cmap='Blues')
```
