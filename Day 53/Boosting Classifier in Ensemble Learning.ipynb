# Importing Dependencies

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import BaggingClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
```

# Data Loading and Preprocesssing

```python
df = sns.load_dataset('iris')
```

# Splitting data into Training and Test Data

```python
X = df.drop(columns='species', axis=1)
Y = df['species']
```

```python
print(X)
```

```python
print(Y)
```

# Label Encoder


```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_Encoder = LabelEncoder()

Y = label_Encoder.fit_transform(Y)
```

```python
print(Y)
```

```python
Y.shape
```

# Train Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=42)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
print(Y.shape, Y_train.shape, Y_test.shape)
```

# 1. AdaBoost

```python
from sklearn.ensemble import AdaBoostClassifier
```

```python
ada_model = AdaBoostClassifier()
```

```python
ada_model.fit(X_train, Y_train)
```

```python
# Prediction

prediction = ada_model.predict(X_test)

acc_score = accuracy_score(Y_test, prediction)

print("Accuracy Score", acc_score)
```

# 2. GradientBoost Classifier

```python
from sklearn.ensemble import GradientBoostingClassifier
```

```python
gradient_boost = GradientBoostingClassifier()
```

```python
gradient_boost.fit(X_train, Y_train)
```

```python
# Prediction 

prediciton = gradient_boost.predict(X_test)

acc_score = accuracy_score(Y_test, prediciton)

print("Accuracy Score: ", acc_score)
```

# 3. XGBoost

```python
from xgboost import XGBClassifier
```

```python
model = XGBClassifier(n_estimator = 100, learning_rate =0.1, max_depth = 3, use_label_encoder = False, eval_metric='mlogloss', random_state=42)
```

```python
model.fit(X_train, Y_train)
```

```python
# Prediction

prediction = model.predict(X_test)

acc_score = accuracy_score(Y_test, prediciton)

print("Accuracy Score:", acc_score)
```
