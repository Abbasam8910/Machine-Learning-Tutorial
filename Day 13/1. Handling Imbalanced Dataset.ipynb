# Distribution of the two classes

```python
credit_data['Class'].value_counts()
```

This is Highly imbalanced Dataset

- 0 --> Legit Transaction
- 1 --> Fraudulent Transaction

# Seperating the legit and fradulent transaction

```python
legit = credit_data[credit_data.Class == 0]
fraud = credit_data[credit_data.Class == 1]
```

```python
print(legit.shape)
```

```python
print(fraud.shape)
```

# Under-Sampling

From 284315 legit transaction will be take 492 legit transaction randomly ( we don't need to take exact we can also take like 500, 450, 520 etc..) - undersampling

 Building a sample dataset containing similar distribution of Legit and Fraudulent transaction 

Number of Fraudulent Transaction -> 492

```python
legit_sample = legit.sample(n=492) # will get 492 random legit samples

```

```python
print(legit_sample.shape)
```

# Concatenate The two DataFrames

```python
new_dataset = pd.concat([legit_sample, fraud], axis = 0) # Axis=0 means we want to concatenate one after the other which means fist entire row of legit_sample will add then fraud
```

```python
new_dataset.tail()
```

Now our classes are evenly distributed
